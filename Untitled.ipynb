- **Biểu đồ tròn (Donut Chart):** Phân tích tỷ trọng thành phần (Movie vs TV Show).
- **Biểu đồ cột ngang (Horizontal Bar Chart):** Xếp hạng thứ tự danh mục (Top 10 Quốc gia, Top 10 Thể loại).
- **Biểu đồ cột phân khúc khán giả (Audience Segments):** Chuẩn hóa và đối chiếu toàn bộ các mức Rating theo độ tuổi và thể loại.

### 1. Bảng số liệu & Biểu đồ tròn (Donut Chart): Tỷ lệ Movie vs TV Show
- **Đối tượng khảo sát:** Cột `type` (Phân loại hình thức phát hành).
- **Biểu đồ lựa chọn:** **Biểu đồ tròn dạng Donut** (tối ưu nhất để biểu diễn tỷ lệ tương quan thành phần trên tổng thể 100%).

```python
# 1. Bảng số liệu chi tiết
type_counts = df_cleaned['type'].value_counts()
total_records = len(df_cleaned)

table_type = pd.DataFrame({
    'Loại hình (Type)': type_counts.index,
    'Số lượng (Titles)': type_counts.values,
    'Tỷ lệ (%)': (type_counts.values / total_records * 100).round(2),
    'Chú thích đối tượng': [
        'Phim lẻ một tập, phim điện ảnh, tài liệu dài',
        'Phim bộ truyền hình, phim dài tập (gồm 1 hoặc nhiều seasons)'
    ]
})

print("=== BẢNG SỐ LIỆU: PHÂN BỐ LOẠI HÌNH NỘI DUNG ===")
display(table_type)

# Biểu diễn bằng BIỂU ĐỒ TRÒN (DONUT CHART)
fig, ax = plt.subplots(figsize=(7, 6))
colors = ['#E50914', '#221F1F']
explode = (0.04, 0)

wedges, texts, autotexts = ax.pie(
    type_counts.values,
    labels=type_counts.index,
    autopct='%1.1f%%',
    pctdistance=0.75,
    startangle=140,
    colors=colors,
    explode=explode,
    textprops=dict(color='black', fontsize=12, fontweight='bold'),
    wedgeprops=dict(width=0.45, edgecolor='white', linewidth=2)
)

for at in autotexts:
    at.set_color('white')
    at.set_fontsize(11)
    at.set_weight('bold')

ax.set_title('Tỷ trọng Nội dung trên Netflix: Phim lẻ (Movie) vs Phim bộ (TV Show)', 
             fontsize=13, fontweight='bold', pad=15)

centre_text = f"Tổng cộng\n{total_records:,}\nNội dung"
ax.text(0, 0, centre_text, ha='center', va='center', fontsize=11, fontweight='bold', color='#333333')

plt.tight_layout()
plt.show()
```

### 4. Kiểm tra & Đối chiếu Số liệu Bảng Phân khúc Khán giả (Audience Segments)
- **Cơ sở đối chiếu với file `netflix_titles_cleaned.csv`:**
  - Cột `rating` chứa các mã phân loại độ tuổi của phim truyền hình và điện ảnh Mỹ.
  - Để phân tích chính xác thị phần đối tượng khán giả, ta quy chuẩn toàn bộ các mã rating về **5 phân khúc khán giả chuẩn hóa**:
    1. **Người trưởng thành (Adults - 18+ / 17+):** Gồm `TV-MA`, `R`, `NC-17`.
    2. **Thanh thiếu niên (Teens - 13+ / 14+):** Gồm `TV-14`, `PG-13`.
    3. **Trẻ lớn / Cần phụ huynh hướng dẫn (Older Kids):** Gồm `TV-PG`, `PG`.
    4. **Trẻ nhỏ & Mọi lứa tuổi (Kids / All Ages):** Gồm `TV-Y7`, `TV-Y`, `TV-G`, `G`, `TV-Y7-FV`.
    5. **Chưa phân loại / Khác (Unrated / Unknown):** Gồm `NR`, `UR`, `Unknown`.
  - Tổng số lượng của 5 phân khúc đảm bảo **khớp 100% với 8.807 dòng** của file dữ liệu.

```python
# 4. Kiểm tra và đối chiếu phân khúc khán giả
def map_audience_segment(r):
    if r in ['TV-MA', 'R', 'NC-17']:
        return '1. Người trưởng thành (Adults 18+)'
    elif r in ['TV-14', 'PG-13']:
        return '2. Thanh thiếu niên (Teens 13-14+)'
    elif r in ['TV-PG', 'PG']:
        return '3. Trẻ lớn / Cần PH hướng dẫn (Older Kids)'
    elif r in ['TV-Y7', 'TV-Y', 'TV-G', 'G', 'TV-Y7-FV']:
        return '4. Trẻ nhỏ & Mọi lứa tuổi (Kids & Family)'
    else:
        return '5. Chưa phân loại / Khác (Unrated / Unknown)'

df_cleaned['audience_segment'] = df_cleaned['rating'].apply(map_audience_segment)

# Bảng đối chiếu số liệu chi tiết
table_audience = df_cleaned.groupby('audience_segment').agg(
    So_luong_Movie=('type', lambda x: (x == 'Movie').sum()),
    So_luong_TVShow=('type', lambda x: (x == 'TV Show').sum()),
    Tong_so_luong=('show_id', 'count')
).reset_index()

table_audience.columns = ['Phân khúc Khán giả', 'Phim lẻ (Movie)', 'Phim bộ (TV Show)', 'Tổng số nội dung']
table_audience['Tỷ lệ (%)'] = (table_audience['Tổng số nội dung'] / total_records * 100).round(2)

segment_details = [
    'Mã: TV-MA (3,207), R (799), NC-17 (3) - Dành riêng khán giả trưởng thành',
    'Mã: TV-14 (2,160), PG-13 (490) - Phù hợp lứa tuổi từ 13, 14 tuổi trở lên',
    'Mã: TV-PG (863), PG (287) - Khuyến cáo định hướng và đồng hành của phụ huynh',
    'Mã: TV-Y7 (334), TV-Y (307), TV-G (220), G (41), TV-Y7-FV (6) - Trẻ em & gia đình',
    'Mã: NR (80), Unknown (7), UR (3) - Chưa phân loại hoặc thiếu dữ liệu'
]
table_audience['Mã Rating tương ứng & Ý nghĩa'] = segment_details

print("=== BẢNG SỐ LIỆU ĐỐI CHIẾU: PHÂN KHÚC KHÁN GIẢ TRÊN NETFLIX ===")
display(table_audience)
print(f"-> Kiểm tra tổng kiểm định: {table_audience['Tổng số nội dung'].sum():,} / {total_records:,} dòng (Khớp 100%).")

# Biểu diễn bằng BIỂU ĐỒ CỘT XẾP CHỒNG (STACKED BAR CHART)
fig, ax = plt.subplots(figsize=(11, 5.5))
plot_data = table_audience.set_index('Phân khúc Khán giả')[['Phim lẻ (Movie)', 'Phim bộ (TV Show)']]
plot_data.plot(kind='barh', stacked=True, color=['#E50914', '#221F1F'], ax=ax, edgecolor='white')

plt.title('Cơ cấu Phân khúc Khán giả theo Loại hình Nội dung (Movie vs TV Show)', fontsize=13, fontweight='bold', pad=15)
plt.xlabel('Số lượng nội dung', fontsize=11)
plt.ylabel('Phân khúc khán giả', fontsize=11)
plt.legend(['Phim lẻ (Movie)', 'Phim bộ (TV Show)'], loc='lower right', frameon=True, facecolor='white')

# Hiển thị tổng số lượng và % ở cuối mỗi thanh
for i, total in enumerate(table_audience['Tổng số nội dung']):
    pct = table_audience.loc[i, 'Tỷ lệ (%)']
    ax.annotate(f" {total:,} ({pct}%)",
                (total, i),
                va='center', fontsize=10, fontweight='bold', color='#111111')

plt.tight_layout()
plt.show()
```

### 5. Kiểm tra & Đối chiếu Số liệu Xu thế Phát triển theo Thời gian
- **Đối chiếu hai cột thời gian cốt lõi trong `netflix_titles_cleaned.csv`:**
  1. **Năm sản xuất / công chiếu gốc (`release_year`):** 
     - Có sẵn nguyên bản trong CSV (kiểu `int64`, không bị thiếu dòng nào).
     - Phản ánh **thời điểm ra đời của tác phẩm**.
     - Đạt đỉnh vào năm **2018** với **1.147** tác phẩm (767 Movie, 380 TV Show).
  2. **Năm đưa lên dịch vụ streaming (`year_added` trích xuất từ `date_added`):**
     - Trích xuất từ chuỗi `date_added` (có 10 dòng bị thiếu `Unknown`).
     - Phản ánh **tốc độ nhập kho và mở rộng danh mục thực tế của Netflix**.
     - Đạt đỉnh vào năm **2019** với **2.016** tác phẩm (1.424 Movie, 592 TV Show).
- **Nguyên nhân có độ lệch giữa hai chỉ số:** Netflix thường mua bản quyền các phim điện ảnh sau khi phim đã chiếu rạp từ 1 đến 2 năm, nên làn sóng phim sản xuất bùng nổ năm 2017-2018 sẽ được cập nhật lên nền tảng tập trung nhất vào năm 2019.

```python
# 5. Đối chiếu số liệu chuỗi thời gian giai đoạn 2012 - 2021
# Thống kê theo release_year
rel_trend = df_cleaned[df_cleaned['release_year'] >= 2012].groupby('release_year')['type'].value_counts().unstack(fill_value=0)
rel_trend['Tong_San_Xuat'] = rel_trend.sum(axis=1)

# Thống kê theo year_added
add_trend = df_cleaned[df_cleaned['year_added'] >= 2012].groupby('year_added')['type'].value_counts().unstack(fill_value=0)
add_trend['Tong_Nhap_Kho'] = add_trend.sum(axis=1)

# Ghép bảng đối chiếu song song
table_comparison = pd.DataFrame({
    'Năm': range(2012, 2022),
    'SX: Movie': [rel_trend.loc[y, 'Movie'] if y in rel_trend.index else 0 for y in range(2012, 2022)],
    'SX: TV Show': [rel_trend.loc[y, 'TV Show'] if y in rel_trend.index else 0 for y in range(2012, 2022)],
    'Năm SX (release_year)': [rel_trend.loc[y, 'Tong_San_Xuat'] if y in rel_trend.index else 0 for y in range(2012, 2022)],
    'Nhập: Movie': [add_trend.loc[y, 'Movie'] if y in add_trend.index else 0 for y in range(2012, 2022)],
    'Nhập: TV Show': [add_trend.loc[y, 'TV Show'] if y in add_trend.index else 0 for y in range(2012, 2022)],
    'Năm Thêm (year_added)': [add_trend.loc[y, 'Tong_Nhap_Kho'] if y in add_trend.index else 0 for y in range(2012, 2022)]
})

print("=== BẢNG SỐ LIỆU ĐỐI CHIẾU: XU THẾ THỜI GIAN (NĂM SẢN XUẤT VS NĂM NHẬP KHO) ===")
display(table_comparison)

# Biểu diễn bằng BIỂU ĐỒ ĐƯỜNG KÉP (DUAL LINE CHART)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 5.5), sharey=False)

years = table_comparison['Năm']

# Đồ thị A: Theo Năm sản xuất gốc (release_year)
ax1.plot(years, table_comparison['Năm SX (release_year)'], marker='o', color='#B20710', linewidth=2.8, label='Tổng SX')
ax1.plot(years, table_comparison['SX: Movie'], marker='s', color='#E50914', linewidth=2, linestyle='--', label='Phim lẻ (Movie)')
ax1.plot(years, table_comparison['SX: TV Show'], marker='^', color='#1E88E5', linewidth=2, linestyle='-.', label='Phim bộ (TV Show)')
ax1.annotate('Đỉnh SX: 1,147 phim\n(Năm 2018)', xy=(2018, 1147), xytext=(2014.5, 1100),
             arrowprops=dict(facecolor='#333333', shrink=0.08, width=1.5, headwidth=6),
             fontweight='bold', fontsize=9.5, color='#B20710')
ax1.set_title('A. Xu hướng Theo Năm Sản Xuất Gốc (release_year)', fontsize=12, fontweight='bold', pad=12)
ax1.set_xlabel('Năm sản xuất (Release Year)', fontsize=11)
ax1.set_ylabel('Số lượng nội dung', fontsize=11)
ax1.set_xticks(years)
ax1.grid(True, linestyle=':', alpha=0.6)
ax1.legend(loc='upper left', frameon=True, facecolor='white')

# Đồ thị B: Theo Năm thêm vào nền tảng (year_added từ date_added)
ax2.plot(years, table_comparison['Năm Thêm (year_added)'], marker='o', color='#B20710', linewidth=2.8, label='Tổng Nhập Kho')
ax2.plot(years, table_comparison['Nhập: Movie'], marker='s', color='#E50914', linewidth=2, linestyle='--', label='Phim lẻ (Movie)')
ax2.plot(years, table_comparison['Nhập: TV Show'], marker='^', color='#1E88E5', linewidth=2, linestyle='-.', label='Phim bộ (TV Show)')
ax2.annotate('Đỉnh Nhập: 2,016 phim\n(Năm 2019)', xy=(2019, 2016), xytext=(2015, 1950),
             arrowprops=dict(facecolor='#333333', shrink=0.08, width=1.5, headwidth=6),
             fontweight='bold', fontsize=9.5, color='#B20710')
ax2.set_title('B. Xu hướng Theo Năm Thêm Vào Nền Tảng (year_added từ date_added)', fontsize=12, fontweight='bold', pad=12)
ax2.set_xlabel('Năm thêm vào Netflix (Year Added)', fontsize=11)
ax2.set_ylabel('Số lượng nội dung', fontsize=11)
ax2.set_xticks(years)
ax2.grid(True, linestyle=':', alpha=0.6)
ax2.legend(loc='upper left', frameon=True, facecolor='white')

plt.tight_layout()
plt.show()
```
