```python
# Import the pandas library
import pandas as pd

# Read the CSV file into a dataframe named 'df'
df = pd.read_csv('netflix_titles.csv')

# Print the number of rows and columns
print(f"Number of rows: {df.shape[0]}")
print(f"Number of columns: {df.shape[1]}\n")

# Display the data type of each column
print("Data types for each column:")
print(df.dtypes)
```

```python
print(f"Tổng số hàng: {df.shape[0]}")
print(f"Tổng số cột: {df.shape[1]}\n")

print("Định dạng dữ liệu của từng cột (Data Types):")
print(df.dtypes)

# 3. Kiểm tra tính đầy đủ của dữ liệu và đếm số lỗi (dữ liệu bị thiếu)
missing_data = df.isnull().sum()
total_errors = missing_data.sum()

print("\nSố lượng dữ liệu bị thiếu (lỗi) ở từng cột:")
print(missing_data[missing_data > 0])
print(f"\n=> Tổng số lần lỗi (ô trống) trong toàn bộ file: {total_errors}")

# 4. Sửa lỗi: Điền thông tin vào các ô trống
# Đối với dữ liệu dạng chữ (object), ta điền 'Unknown' hoặc 'Chưa xác định'
df.fillna('Unknown', inplace=True)
df.to_csv('netflix_titles_cleaned.csv', index=False)

# (Tùy chọn) Nếu có cột dạng số bị thiếu, bạn có thể điền số 0, ví dụ:
# df_cleaned['Tên_cột_số'] = df['Tên_cột_số'].fillna(0)

print("\nĐã hoàn tất việc sửa lỗi bằng cách lấp đầy các ô trống!")
```
