## 5. TRỰC QUAN HÓA CÁC LỖI VÀ DỮ LIỆU BỊ THIẾU
Phần này biểu diễn số liệu chi tiết về 4.307 ô dữ liệu bị thiếu (lỗi) ở 6 cột trước khi được xử lý bằng cách điền `'Unknown'` thông qua duy nhất 1 biểu đồ Donut chi tiết.

```python
# Cấu hình biểu đồ với font hỗ trợ tiếng Việt
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

plt.rcParams['font.family'] = 'Segoe UI'
plt.rcParams['figure.dpi'] = 100
sns.set_theme(style="whitegrid", font='Segoe UI')

# 5.1 Biểu đồ Donut chi tiết: Hiển thị đầy đủ 6/6 cột dữ liệu bị thiếu (kèm callouts cho 3 lát cắt nhỏ)
missing_counts = missing_data[missing_data > 0].sort_values(ascending=False)
total_errors = missing_counts.sum()
total_rows = len(df)

# Màu sắc phân biệt trực quan cho cả 6 lát
colors = ['#E50914', '#FF7043', '#FFA726', '#00897B', '#1E88E5', '#8E24AA']

# Tách nhẹ (explode) 3 lát cắt nhỏ ở cuối để chúng lộ rõ trên vòng Donut
explode = [0, 0, 0, 0.08, 0.16, 0.24]

fig, ax = plt.subplots(figsize=(13.5, 7.5), subplot_kw=dict(aspect='equal'))

wedges, texts = ax.pie(
    missing_counts,
    explode=explode,
    startangle=140,
    colors=colors,
    wedgeprops=dict(width=0.45, edgecolor='black', linewidth=1.2)
)

# Hiển thị thông tin tổng hợp ở tâm biểu đồ Donut
ax.text(0, 0.12, 'TỔNG SỐ LỖI', ha='center', va='center', fontsize=12, fontweight='bold', color='#555555')
ax.text(0, -0.05, f'{total_errors:,}', ha='center', va='center', fontsize=26, fontweight='bold', color='#E50914')
ax.text(0, -0.18, 'ô trống bị thiếu', ha='center', va='center', fontsize=11, color='#666666')

# 1. Nhãn số liệu trực tiếp bên trong 3 lát cắt lớn
labels_large = {
    0: f'director\n2,634 ô\n(61.16%)',
    1: f'country\n831 ô\n(19.29%)',
    2: f'cast\n825 ô\n(19.15%)'
}

for i in range(3):
    w = wedges[i]
    theta = np.deg2rad((w.theta1 + w.theta2) / 2)
    r = 0.77
    x = r * np.cos(theta)
    y = r * np.sin(theta)
    ax.text(x, y, labels_large[i], ha='center', va='center', fontsize=11, fontweight='bold', color='white')

# 2. Mũi tên chỉ dẫn (Callout leader lines) hiển thị chi tiết 3 số liệu nhỏ ở cuối
callouts = [
    (3, 'date_added', (-1.45, 1.05)),
    (4, 'rating',     (-1.60, 0.70)),
    (5, 'duration',   (-1.45, 0.35))
]

for idx, col_name, (tx, ty) in callouts:
    w = wedges[idx]
    val = missing_counts[col_name]
    pct = (val / total_errors) * 100
    
    theta = np.deg2rad((w.theta1 + w.theta2) / 2)
    exp = explode[idx]
    x_pt = (1.0 + exp) * np.cos(theta)
    y_pt = (1.0 + exp) * np.sin(theta)
    
    label_str = f'{col_name}: {val} ô ({pct:.2f}% lỗi)'
    
    ax.annotate(
        label_str,
        xy=(x_pt, y_pt),
        xytext=(tx, ty),
        arrowprops=dict(arrowstyle='->', connectionstyle='arc3,rad=-0.1', color=colors[idx], lw=2),
        fontsize=10.5,
        fontweight='bold',
        color=colors[idx],
        bbox=dict(boxstyle='round,pad=0.35', facecolor='#ffffff', edgecolor=colors[idx], lw=1.5),
        ha='right',
        va='center'
    )

# 3. Bảng Legend thống kê chi tiết đặt bên phải
legend_labels = [
    f'{col.ljust(12)}: {val:>5,} ô ({val/total_errors*100:>5.2f}% tổng lỗi | {val/total_rows*100:>5.2f}% tập dữ liệu)'
    for col, val in zip(missing_counts.index, missing_counts.values)
]

ax.legend(
    wedges, 
    legend_labels,
    title='BẢNG THỐNG KÊ CHI TIẾT 6 CỘT',
    title_fontsize='12',
    loc='center left',
    bbox_to_anchor=(1.05, 0.5),
    fontsize=10.5,
    prop={'family': 'Consolas', 'size': 10},
    frameon=True,
    facecolor='#fdfdfd',
    edgecolor='#cccccc'
)

ax.set_title('Biểu đồ Donut: Phân bố chi tiết 4,307 dữ liệu bị thiếu (Đầy đủ nhãn 6/6 cột)', fontsize=15, fontweight='bold', pad=25)
plt.tight_layout()
plt.show()
```

> * **Tổng số lỗi ô trống**: Có **4.307** ô dữ liệu bị thiếu trên toàn bộ 8.807 dòng dữ liệu gốc.
> * **Các cột lỗi trọng điểm**:
>   * `director` chiếm đa số với **2.634 ô** (chiếm **61.16%** tổng số lỗi, tương đương **29.91%** toàn bộ tập dữ liệu).
>   * `country` thiếu **831 ô** (chiếm **19.29%** tổng lỗi, tương đương **9.44%** tập dữ liệu).
>   * `cast` thiếu **825 ô** (chiếm **19.15%** tổng lỗi, tương đương **9.37%** tập dữ liệu).
>   * Ba cột trên chiếm tới **99.6%** tổng số ô lỗi trong toàn file.
> * **Các cột lỗi rất ít**: `date_added` (10 ô ~ 0.11%), `rating` (4 ô ~ 0.05%), `duration` (3 ô ~ 0.03%).
> * **Định hướng CSDL**: Không nên ép ràng buộc `NOT NULL` cho các cột `director`, `cast`, `country`. Trong mô hình quan hệ, nên cho phép `NULL` hoặc liên kết với ID bản ghi mặc định (*'Chưa xác định'*).

## 6. PHÂN TÍCH VÀ TRỰC QUAN HÓA DỮ LIỆU TỪ `netflix_titles_cleaned.csv`

```python
# Đọc lại file dữ liệu đã được làm sạch
df_cleaned = pd.read_csv('netflix_titles_cleaned.csv')

# 6.1 Biểu đồ cột: Phân bố số lượng và tỷ lệ loại hình nội dung (Movie vs TV Show)
type_counts = df_cleaned['type'].value_counts()
total_titles = len(df_cleaned)

plt.figure(figsize=(9, 6))
colors = ['#E50914', '#221F1F']  # Đỏ Netflix và Đen

bars = plt.bar(type_counts.index, type_counts.values, color=colors, edgecolor='black', width=0.52, alpha=0.9)
plt.title("Phân bố số lượng và tỷ lệ nội dung: Phim (Movie) vs TV Show", fontsize=14, fontweight='bold', pad=20)
plt.xlabel("Loại hình nội dung", fontsize=12, fontweight='bold')
plt.ylabel("Số lượng tựa phim / show", fontsize=12, fontweight='bold')
plt.ylim(0, 7200)

# Hiển thị số lượng và tỷ lệ % trực tiếp trên đầu mỗi cột
for bar in bars:
    yval = bar.get_height()
    pct = (yval / total_titles) * 100
    plt.text(
        bar.get_x() + bar.get_width() / 2, 
        yval + 140, 
        f"{int(yval):,} tác phẩm\n({pct:.1f}%)", 
        ha='center', 
        va='bottom', 
        fontsize=12, 
        fontweight='bold'
    )

plt.tight_layout()
plt.show()
```

```python
# 6.3 Xu hướng phát hành nội dung theo năm: Phim (Movie) vs TV Show (Giai đoạn 2000 - 2021)
# Biểu diễn chuẩn xác số liệu từng năm và phân chia các giai đoạn phát triển rõ ràng
df_recent = df_cleaned[df_cleaned['release_year'] >= 2000]
trend_data = df_recent.groupby(['release_year', 'type']).size().unstack(fill_value=0)

fig, ax = plt.subplots(figsize=(16, 8))

# Tô màu phân chia 4 giai đoạn phát triển rõ rệt của Netflix
ax.axvspan(1999.5, 2010.5, color='#F5F5F5', alpha=0.7, label='Giai đoạn 1 (2000-2010): Tiền đề & Tích lũy')
ax.axvspan(2010.5, 2015.5, color='#E8F5E9', alpha=0.55, label='Giai đoạn 2 (2011-2015): Tăng tốc Streaming')
ax.axvspan(2015.5, 2018.5, color='#FFF3E0', alpha=0.65, label='Giai đoạn 3 (2016-2018): Bùng nổ đỉnh cao')
ax.axvspan(2018.5, 2021.5, color='#EDE7F6', alpha=0.6, label='Giai đoạn 4 (2019-2021): TV Show bứt phá & Đại dịch')

# Vẽ 2 đường xu hướng với điểm đánh dấu rõ ràng
line_movie = ax.plot(trend_data.index, trend_data['Movie'], marker='o', markersize=6.5, color='#E50914', linewidth=2.5, label='Movie (Phim lẻ)', zorder=4)
line_tv = ax.plot(trend_data.index, trend_data['TV Show'], marker='s', markersize=6.5, color='#1E88E5', linewidth=2.5, label='TV Show (Truyền hình)', zorder=4)

# Hiển thị số liệu chuẩn xác tại TỪNG NĂM cho cả Movie và TV Show
for yr in trend_data.index:
    m = trend_data.loc[yr, 'Movie']
    t = trend_data.loc[yr, 'TV Show']
    
    if m >= t:
        # Movie nằm trên, TV Show nằm dưới
        ax.text(yr, m + 18, f'{m}', ha='center', va='bottom', fontsize=8.5, fontweight='bold', color='#B71C1C', zorder=5)
        ax.text(yr, t - 22, f'{t}', ha='center', va='top', fontsize=8.5, fontweight='bold', color='#0D47A1', zorder=5)
    else:
        # TV Show nằm trên (như năm 2021), Movie nằm dưới
        ax.text(yr, t + 18, f'{t}', ha='center', va='bottom', fontsize=8.5, fontweight='bold', color='#0D47A1', zorder=5)
        ax.text(yr, m - 22, f'{m}', ha='center', va='top', fontsize=8.5, fontweight='bold', color='#B71C1C', zorder=5)

# Gắn nhãn các mốc phát triển quan trọng (Milestones)
ax.annotate('Đỉnh cao Movie:\n767 phim (2017 & 2018)', xy=(2017.5, 767), xytext=(2014.8, 815),
            arrowprops=dict(arrowstyle='->', color='#E50914', lw=1.8),
            fontsize=10, fontweight='bold', color='#B71C1C',
            bbox=dict(boxstyle='round,pad=0.35', facecolor='#FFEBEE', edgecolor='#E50914', lw=1.2), zorder=6)

ax.annotate('Đỉnh cao TV Show:\n436 show (2020)', xy=(2020, 436), xytext=(2018.6, 520),
            arrowprops=dict(arrowstyle='->', color='#1E88E5', lw=1.8),
            fontsize=10, fontweight='bold', color='#0D47A1',
            bbox=dict(boxstyle='round,pad=0.35', facecolor='#E3F2FD', edgecolor='#1E88E5', lw=1.2), zorder=6)

ax.annotate('Năm 2021:\nTV Show (315) > Movie (277)', xy=(2021, 315), xytext=(2019.5, 230),
            arrowprops=dict(arrowstyle='->', color='#4A148C', lw=1.6),
            fontsize=9.5, fontweight='bold', color='#4A148C',
            bbox=dict(boxstyle='round,pad=0.3', facecolor='#F3E5F5', edgecolor='#4A148C', lw=1), zorder=6)

ax.set_ylim(-35, 890)
ax.set_xlim(1999.3, 2021.7)
ax.set_xticks(trend_data.index)
ax.tick_params(axis='x', rotation=45, labelsize=10)
ax.tick_params(axis='y', labelsize=10)

ax.set_title("Xu hướng phát hành nội dung: Phim (Movie) vs TV Show theo từng năm (2000 - 2021)", fontsize=15, fontweight='bold', pad=22)
ax.set_xlabel("Năm phát hành (Release Year)", fontsize=12, fontweight='bold')
ax.set_ylabel("Số lượng tác phẩm phát hành", fontsize=12, fontweight='bold')
ax.grid(True, linestyle='--', alpha=0.5)

# Bảng chú thích (Legend)
ax.legend(loc='upper left', fontsize=10.5, framealpha=0.92, facecolor='#ffffff', edgecolor='#cccccc')

plt.tight_layout()
plt.show()
```

```python
# 6.4 Phân bố xếp hạng độ tuổi (Content Rating): Movie vs TV Show
# Xử lý dữ liệu: Loại bỏ các giá trị bị lỗi lệch cột ('74 min', '84 min', '66 min') và 'Unknown'
invalid_ratings = ['74 min', '84 min', '66 min', 'Unknown']
df_rating = df_cleaned[~df_cleaned['rating'].isin(invalid_ratings)].copy()

# Sắp xếp thứ tự các bậc xếp hạng phổ biến nhất
rating_order = ['TV-MA', 'TV-14', 'TV-PG', 'R', 'PG-13', 'TV-Y7', 'TV-Y', 'PG', 'TV-G', 'NR', 'G']
df_rating_top = df_rating[df_rating['rating'].isin(rating_order)]

fig, ax = plt.subplots(figsize=(16, 7.5))

sns.countplot(
    data=df_rating_top, 
    x='rating', 
    order=rating_order, 
    hue='type', 
    palette=['#E50914', '#1E88E5'], 
    edgecolor='black',
    ax=ax
)

# Hiển thị số liệu chuẩn xác trên TỪNG CỘT cho cả Movie và TV Show
for p in ax.patches:
    height = p.get_height()
    if not np.isnan(height) and height > 0:
        ax.annotate(
            f'{int(height):,}',
            (p.get_x() + p.get_width() / 2., height),
            ha='center', va='bottom',
            fontsize=9.5, fontweight='bold',
            xytext=(0, 4), textcoords='offset points'
        )

ax.set_ylim(0, 2300)
ax.set_title("Phân bố xếp hạng độ tuổi (Content Ratings): So sánh Phim (Movie) và TV Show", fontsize=15, fontweight='bold', pad=22)
ax.set_xlabel("Mã xếp hạng độ tuổi (Rating Code)", fontsize=12, fontweight='bold')
ax.set_ylabel("Số lượng tác phẩm", fontsize=12, fontweight='bold')
ax.legend(title="Loại hình", title_fontsize='11.5', fontsize=11, loc='upper right', framealpha=0.95)
ax.grid(True, linestyle='--', alpha=0.5, axis='y')

plt.tight_layout()
plt.show()
```

> 📌 **Ghi chú & Nhận xét Biểu đồ 6.4 (Note) - Chi tiết từng độ tuổi:**
>

```python
# 6.6 Phân bố thời lượng của Phim lẻ (Movie Duration Distribution)
movie_df = df_cleaned[(df_cleaned['type'] == 'Movie') & (df_cleaned['duration'] != 'Unknown')].copy()
movie_df['duration_min'] = movie_df['duration'].str.extract(r'(\d+)').astype(float)

plt.figure(figsize=(13, 6))
sns.histplot(movie_df['duration_min'].dropna(), kde=True, color='#E50914', bins=35, edgecolor='black', alpha=0.6)
mean_dur = movie_df['duration_min'].mean()
median_dur = movie_df['duration_min'].median()

plt.axvline(mean_dur, color='black', linestyle='--', linewidth=2, label=f"Thời lượng trung bình: {mean_dur:.1f} phút")
plt.axvline(median_dur, color='blue', linestyle=':', linewidth=2, label=f"Thời lượng trung vị: {median_dur:.1f} phút")
plt.title("Phân bố thời lượng các bộ phim lẻ trên Netflix (Movie Duration in Minutes)", fontsize=14, fontweight='bold', pad=15)
plt.xlabel("Thời lượng (Phút)", fontsize=12)
plt.ylabel("Số lượng phim", fontsize=12)
plt.legend(fontsize=12)
plt.tight_layout()
plt.show()
```

> * Một số ít phim tài liệu ngắn dưới 40 phút hoặc phim điện ảnh dài trên 180 phút (3 tiếng).

## 7. TỔNG KẾT VÀ ĐỀ XUẤT THIẾT KẾ CƠ SỞ DỮ LIỆU (DATABASE INSIGHTS)

Tổng hợp các phát hiện từ quá trình EDA để áp dụng trực tiếp vào thiết kế CSDL quan hệ:

1. **Chuẩn hóa dữ liệu (Normalization - 3NF)**:
   - Các trường đa trị (`cast`, `director`, `country`, `listed_in`) cần được phân tách thành các bảng riêng: `Person` / `Actor` / `Director`, `Country`, `Category`.
   - Sử dụng các bảng trung gian để liên kết nhiều-nhiều: `Title_Director`, `Title_Cast`, `Title_Country`, `Title_Category`.
2. **Kiểu dữ liệu phù hợp (Data Types)**:
   - `show_id`: Khóa chính `VARCHAR(10)` (Primary Key).
   - `release_year`: `SMALLINT`.
   - `date_added`: Chuẩn hóa về kiểu `DATE` (YYYY-MM-DD).
   - `duration`: Tách thành `duration_value` (INT) và `duration_unit` (VARCHAR).
3. **Xử lý giá trị NULL**:
   - `director`, `country`, `cast` có tỷ lệ thiếu từ 9% đến 30%, do đó trong CSDL các trường quan hệ này cần cho phép giá trị `NULL` hoặc gán ID mặc định đại diện cho *'Chưa xác định'*.
